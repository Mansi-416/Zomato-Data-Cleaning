```python
import numpy as np
import pandas as pd
```

```python
#load the dataset
df=pd.read_csv("Zomato Restaurant dataset.csv")
print("Dataset loaded successfully!")
print("Dataset shape:", df.shape)
display(df.head())
```

```python
#Inspect the dataset
print("Dataset information:")
print(df.info())
print("\nColumn names:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
```

```python
#show the statistic for numerical columns in the dataset
print(df.describe())
```

```python
#check for missing values in the dataset
print("Missing values in each column:")
print(df.isnull().sum())
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
#prints the columns that only containing missing values
missing = df.isnull().sum()

print("Columns containing missing values:")
print(missing[missing > 0])

#handle missing values
df["Cuisines"]=df["Cuisines"].fillna("Not Specified")
print("Missing Cuisines:", df["Cuisines"].isnull().sum())
```

```python
#check for duplicate rows and drop them from the dataset
print("Duplicate Records:", df.duplicated().sum())
print("Total number of duplicate values:", df.duplicated().sum().sum())
df=df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

```python
#clean inconsistent text values
text_columns=df.select_dtypes(include=["str","object"]).columns
for col in text_columns:
    df[col]=df[col].str.strip()

#covert blank strings into missing values
df.replace(r"^\s*$",np.nan,regex=True,inplace=True)
    
```

```python
#Identify and correct data types
print(df.dtypes)

numeric_columns = [
    "Restaurant ID",
    "Country Code",
    "Longitude",
    "Latitude",
    "Average Cost for two",
    "Price range",
    "Aggregate rating",
    "Votes"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print("\nData types after conversion:")
print(df[numeric_columns].dtypes)
```

```python
#standardize inconsistent Yes/No values
boolean_columns=[
    "Has Online delivery",
    "Is delivering now",
    "Switch to order menu",
    "Has Table booking"
]
for col in boolean_columns:
    print(f"\n{col}:")
    df[col]=df[col].replace({
        0:"No",
        1:"Yes"
    })
    print("After cleaning: ")
    print(df[col].value_counts(dropna=False).rename_axis(None))
```

```python
#check inconsistent categorical values
categorical_columns=[
    "Country Code",
    "City",
    "Currency",
    "Rating text",
    "Has Table booking",
    "Has Online delivery"
]
for col in categorical_columns:
    if col in df.columns:
        print(f"\nUniques Values in {col}:")
        print(df[col].value_counts(dropna=False))
    else:
        print(f"\n Column not found: {col}")
```

```python
#add rating status column
df["Rating Status"]=np.where(
    df["Rating text"].eq("Not rated"),
    "Not Rated",
    "Rated"
)
print(df["Rating Status"].value_counts())
```

```python
#validate aggregate ratings
invalid_ratings = df[
    ~df["Aggregate rating"].between(0, 5)
    & df["Aggregate rating"].notna()
]

print("Invalid ratings:", len(invalid_ratings))
```

```python
#validate price ranges
invalid_prices = df[
    ~df["Price range"].between(1, 4)
    & df["Price range"].notna()
]

print("Invalid price ranges:", len(invalid_prices))
```

```python
#check for negative ratings and price
invalid_costs = df[
    df["Average Cost for two"].lt(0)
    & df["Average Cost for two"].notna()
]

invalid_votes = df[
    df["Votes"].lt(0)
    & df["Votes"].notna()
]

print("Negative restaurant costs:", len(invalid_costs))
print("Negative vote counts:", len(invalid_votes))
```

```python
#check missing values after cleaning
print("Missing values after cleaning:")
print(df.isnull().sum())
print("Total missing cells:", df.isnull().sum().sum())
```

```python
#check duplicates again
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
#final validation
print("========== FINAL DATASET SUMMARY ==========")

print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])

print("Duplicate records:", df.duplicated().sum())

print("Total missing values:", df.isnull().sum().sum())

print("\nFinal data types:")
print(df.dtypes)

print("\nFirst five rows:")
display(df.head())
```

```python
df.to_csv(
    "Zomato_Restaurant_Dataset_Cleaned.csv",
    index=False
)

print("Cleaned dataset saved successfully!")
```
